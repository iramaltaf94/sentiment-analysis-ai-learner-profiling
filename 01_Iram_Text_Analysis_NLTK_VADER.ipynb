# **Learner Sentiment, Confidence & Cognitive Load Analysis**

### **NLTK VADER text analytics on mentor session transcripts**

This notebook is a **reusable function**: upload **one CSV** with a row per learner, and it analyses every learner automatically.

**This run covers 18 learners** whose session transcripts were available. Learner names are anonymized as **Student 1, Student 2, ...** to protect identities, and all scores shown are real results from the data.

**What it measures**
- **Sentiment (NLTK VADER):** compound score from -1 to +1. Positive if ≥ 0.05, negative if ≤ -0.05, neutral in between.
- **Confidence score (out of 10):** (high-confidence keywords found ÷ all keywords found) × 10. If no keyword is found, the score is a neutral 5.0.
- **Cognitive load:** counted from hesitation words (*uh, umm, actually, like, sorry, for example*). Low: 0–2, Moderate: 3–5, High: more than 5.

**Input CSV columns:** `name`, `session`, `text`, `high_conf`, `low_conf` (keywords separated by `;`). To analyse another batch, upload a new CSV with the same columns.

```python
# ===== Learner Sentiment, Confidence & Cognitive Load Analysis (NLTK VADER) =====
# Upload ONE CSV with columns: name, session, text, high_conf, low_conf
# (keywords separated by ';'). Every learner in the CSV is analysed automatically.

import nltk
import pandas as pd
import matplotlib.pyplot as plt
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from google.colab import files

nltk.download('vader_lexicon', quiet=True)
sia = SentimentIntensityAnalyzer()

HESITATION = ["uh", "umm", "actually", "like", "sorry", "for example"]


def analyze_student(name, session, text, high_conf, low_conf, analyst="Iram", plot=True):
    """Analyse one learner's transcript and return the scores as a dictionary."""
    t = text.lower()
    high_count = sum(1 for w in high_conf if w in t)
    low_count = sum(1 for w in low_conf if w in t)
    total = high_count + low_count
    confidence_score = round((high_count / total) * 10, 1) if total > 0 else 5.0

    hes_count = sum(t.count(w) for w in HESITATION)
    cognitive_load = "High" if hes_count > 5 else "Moderate" if hes_count > 2 else "Low"

    sentiment = sia.polarity_scores(text)
    overall = ("Positive" if sentiment['compound'] >= 0.05
               else "Negative" if sentiment['compound'] <= -0.05 else "Neutral")

    print("=" * 42)
    print(f"  {name.upper()} — INTERACTION ANALYSIS")
    print(f"  Session : {session}")
    print(f"  Analyst : {analyst}")
    print("=" * 42)
    print(f"  Confidence Score : {confidence_score}/10")
    print(f"  Cognitive Load   : {cognitive_load} ({hes_count} hesitation words)")
    print(f"  Sentiment        : {overall} ({round(sentiment['compound'], 2)})")
    print(f"  Positive Language: {round(sentiment['pos'] * 100)}%")
    print(f"  Negative Language: {round(sentiment['neg'] * 100)}%")
    print("=" * 42)

    if plot:
        fig, axes = plt.subplots(1, 3, figsize=(13, 4))
        fig.suptitle(f"{name} — Interaction Analysis", fontsize=13, fontweight='bold')
        axes[0].bar(["High", "Low"], [high_count, low_count], color=["#2ecc71", "#e74c3c"], width=0.4)
        axes[0].set_title(f"Confidence\n{confidence_score}/10")
        axes[0].set_ylabel("Word Count")
        axes[1].bar(["Hesitation Words"], [hes_count], color=["#e67e22"], width=0.4)
        axes[1].set_title(f"Cognitive Load\n{cognitive_load}")
        axes[1].set_ylabel("Count")
        axes[2].bar(["Positive", "Negative", "Neutral"],
                    [sentiment['pos'], sentiment['neg'], sentiment['neu']],
                    color=["#2ecc71", "#e74c3c", "#3498db"], width=0.4)
        axes[2].set_title(f"Sentiment\n{overall}")
        axes[2].set_ylabel("Score")
        plt.tight_layout()
        plt.show()

    return {"name": name, "session": session, "confidence": confidence_score,
            "hesitation_words": hes_count, "cognitive_load": cognitive_load,
            "sentiment": overall, "compound": round(sentiment['compound'], 2)}


def analyze_csv(path, plot=True):
    """Run analyze_student for every row of the CSV and return a summary table."""
    df = pd.read_csv(path)
    results = []
    for _, row in df.iterrows():
        high = [w.strip() for w in str(row["high_conf"]).split(";") if w.strip()]
        low = [w.strip() for w in str(row["low_conf"]).split(";") if w.strip()]
        results.append(analyze_student(row["name"], row["session"], str(row["text"]), high, low, plot=plot))
    return pd.DataFrame(results)


# ---- Upload the CSV and run ----
uploaded = files.upload()
CSV_PATH = list(uploaded.keys())[0]
summary = analyze_csv(CSV_PATH)
display(summary)

# ---- Overall sentiment of all learners ----
counts = summary["sentiment"].value_counts().reindex(["Positive", "Neutral", "Negative"], fill_value=0)
ax = counts.plot(kind="bar", color=["#0A9396", "#94A3B8", "#E63946"], figsize=(6, 4), rot=0, width=0.4)
ax.set_title(f"Sentiment of {len(summary)} learners (NLTK VADER)")
ax.set_ylabel("Number of learners")
for i, v in enumerate(counts.values):
    ax.text(i, v + 0.2, str(v), ha="center", fontweight="bold")
plt.show()
```
